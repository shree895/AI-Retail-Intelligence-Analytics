```python
df.head()
```

```python
customer_spend = df.groupby('customer_unique_id')['Revenue'].sum().reset_index()
```

```python
customer_spend.head()
```

# Customer Segmentation

```python
from sklearn.cluster import KMeans
```

```python
X = customer_spend[['Revenue']]
```

```python
kmeans = KMeans(n_clusters = 3 , random_state = 42)
```

```python
customer_spend['Segment'] = kmeans.fit_predict(X)
```

```python
customer_spend.head()
```

```python
mapping = {
    0 : 'Low Value' , 
    1 : 'Medium Value' ,
    2 : 'High Value' 
}
customer_spend['Segment'] = customer_spend['Segment'].map(mapping)
```

```python
customer_spend.head()
```

```python
customer_spend.to_csv(r"C:\Users\SHREE SINHA\OneDrive\Desktop\AI_Retail_Intelligence_Project\Data_cleaned\customer_segments.csv", index = False)
```

```python
import numpy as np
```

```python
customer_spend['Log_Revenue'] = np.log1p(
    customer_spend['Revenue']
)
```

```python
X = customer_spend[['Log_Revenue']]
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42
)
```

```python
customer_spend['Cluster'] = kmeans.fit_predict(X)
```

```python
customer_spend.groupby('Cluster')['Revenue'].mean()
```

```python
customer_spend['Cluster'].value_counts()
```

```python
mapping = {
    0: 'Low Value',
    1: 'Medium Value',
    2: 'High Value'
}

customer_spend['Segment'] = customer_spend[
    'Cluster'
].map(mapping)
```
